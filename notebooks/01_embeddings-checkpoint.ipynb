```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

print("Embedding model loaded successfully!")
```

```python
documents = [
    "JES2 spool utilization is increasing due to retained job output.",
    "High JES2 spool usage can delay batch job processing.",
    "Batch workload performance is affected by resource contention.",
    "JES2 spool cleanup helps reclaim disk capacity.",
    "CICS transaction response time is increasing.",
    "Db2 query optimization improves database performance.",
    "TCP/IP network configuration on z/OS.",
    "Mainframe CPU utilization is high during peak workload.",
    "Storage shortages can cause batch job failures.",
    "Use SDSF to monitor JES2 spool and job output."
]

print("Total documents:", len(documents))
```

```python
embeddings = model.encode(
    documents,
    normalize_embeddings=True
)

print("Embedding shape:", embeddings.shape)
print("First document embedding:", embeddings[0][:10])
```

```python
similarity_matrix = embeddings @ embeddings.T

df = pd.DataFrame(
    similarity_matrix,
    index=[f"D{i+1}" for i in range(len(documents))],
    columns=[f"D{i+1}" for i in range(len(documents))]
)

display(df.round(3))
```

```python
plt.figure(figsize=(9, 7))
plt.imshow(similarity_matrix, cmap="viridis", vmin=-1, vmax=1)
plt.colorbar(label="Cosine similarity")
plt.xticks(range(len(documents)), [f"D{i+1}" for i in range(len(documents))])
plt.yticks(range(len(documents)), [f"D{i+1}" for i in range(len(documents))])
plt.title("Document-to-Document Semantic Similarity")
plt.show()
```

```python
def semantic_search(query, top_k=5):
    query_embedding = model.encode(
        query,
        normalize_embeddings=True
    )

    scores = embeddings @ query_embedding

    top_indices = np.argsort(scores)[::-1][:top_k]

    return pd.DataFrame([
        {
            "rank": rank,
            "document_id": f"D{idx+1}",
            "score": round(float(scores[idx]), 4),
            "document": documents[idx]
        }
        for rank, idx in enumerate(top_indices, start=1)
    ])

semantic_search("Why are my batch jobs running slowly?")
```

```python
queries = [
    "Why are my batch jobs running slowly?",
    "How do I reduce JES2 spool utilization?",
    "How can I improve database query performance?",
    "Why is my application network slow?",
    "What causes high CPU utilization?"
]

for query in queries:
    print(f"\nQuery: {query}")
    display(semantic_search(query, top_k=3))
```

```python
query = "Why is JES2 spool utilization high?"

q = model.encode(query, normalize_embeddings=True)

cosine_scores = embeddings @ q

euclidean_distances = np.linalg.norm(
    embeddings - q,
    axis=1
)

comparison = pd.DataFrame({
    "document": documents,
    "cosine_similarity": cosine_scores,
    "euclidean_distance": euclidean_distances
})

display(
    comparison.sort_values(
        "cosine_similarity",
        ascending=False
    ).round(4)
)
```

```python
def evaluate_retrieval(query, relevant_ids, k=3):
    results = semantic_search(query, top_k=k)

    retrieved = set(results["document_id"])
    relevant = set(relevant_ids)

    correct = len(retrieved & relevant)

    precision = correct / k
    recall = correct / len(relevant) if relevant else None

    return {
        "precision_at_k": precision,
        "recall_at_k": recall,
        "retrieved_ids": list(results["document_id"]),
    }

evaluate_retrieval(
    "Why are my batch jobs running slowly?",
    relevant_ids={"D2", "D3"},
    k=3
)
```
