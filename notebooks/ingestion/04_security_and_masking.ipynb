# 04 — Security, masking, and access controls
**Goal:** Detect sensitive-looking strings and produce a safe copy when policy requires it.

**Must:** preserve document ACL/security classification and protect credentials.
**Conditional:** PII redaction, pseudonymization, manual review.

**Trade-off:** masking reduces exposure but may harm answering questions involving the masked identity. Regex is insufficient for complete PII detection.

Use this notebook with synthetic demo data only.

```python
import re
source = '''Contact admin@example.com for JES2 troubleshooting.
Ticket owner: person@example.org. API key: sk-demo-not-a-real-secret
Command: $D SPOOL
'''

def mask_demo(text):
    emails = re.findall(r'(?i)\b[A-Z0-9._%+-]+@[A-Z0-9.-]+\.[A-Z]{2,}\b', text)
    out = re.sub(r'(?i)\b[A-Z0-9._%+-]+@[A-Z0-9.-]+\.[A-Z]{2,}\b', '<EMAIL>', text)
    out = re.sub(r'\bsk-demo-[A-Za-z0-9-]+\b', '<DEMO_TOKEN>', out)
    return out, emails

masked, emails = mask_demo(source)
print(masked)
print('email count:', len(emails))
assert '$D SPOOL' in masked
assert 'admin@example.com' not in masked
```
