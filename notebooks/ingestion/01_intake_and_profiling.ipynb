# 01 — Intake and document profiling
**Goal:** Decide *what is this file*, whether to accept it, and which parser it probably needs.

**Must:** file existence, type, size limits, stable ID/hash, source metadata, trust boundary.
**Trade-off:** extension sniffing is cheap but less reliable than content inspection. A checksum finds byte changes, not semantic changes.

**Output:** a small profiling dictionary. **No embeddings yet.**

```python
from pathlib import Path
import hashlib

DATA = Path('../data') if Path('../data').exists() else Path('data')
pdf = DATA / 'jes2_manual.pdf'
assert pdf.is_file(), f'Missing {pdf}'
raw = pdf.read_bytes()
profile = {
    'name': pdf.name,
    'suffix': pdf.suffix.lower(),
    'bytes': len(raw),
    'sha256': hashlib.sha256(raw).hexdigest(),
    'pdf_magic_header': raw.startswith(b'%PDF-'),
}
profile
```
