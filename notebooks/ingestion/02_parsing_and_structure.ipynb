# 02 — Extraction, parsing, and structure
**Goal:** Compare extracting flat text with retaining coordinates, blocks and tables.

**Must:** recover useful content and identify where it came from. **Conditional:** OCR for image-based pages, layout models for complex layouts.

**Trade-off:** native extraction is fast; structural reconstruction is more accurate for tables but costs more.

Run: `uv pip install pymupdf` if needed.

```python
from pathlib import Path
import pymupdf
DATA = Path('../data') if Path('../data').exists() else Path('data')
doc = pymupdf.open(DATA / 'jes2_manual.pdf')
for i, page in enumerate(doc, 1):
    text = page.get_text('text', sort=True)
    print(f'PAGE {i}\n{text[:900]}')
```

```python
page = doc[0]
blocks = page.get_text('blocks', sort=True)
for b in blocks[:8]:
    print({'bbox': tuple(round(v,1) for v in b[:4]), 'text': b[4][:90].replace('\n',' | ')})

tables = page.find_tables()
print('Tables detected:', len(tables.tables))
for table in tables.tables:
    for row in table.extract():
        print(row)
```
